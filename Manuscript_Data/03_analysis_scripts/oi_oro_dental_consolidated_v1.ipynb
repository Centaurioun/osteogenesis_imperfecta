# OI Analiz Notebook Oluşturma Akışı (v1)

Bu notebook, projeye uygun adlandırmayla (`*_v1.ipynb`) Jupyter notebook dosyasını **programatik** üretme akışını adım adım gösterir.

---
## 1) Proje Adını ve Notebook Dosya Adını Tanımlama (`*_v1.ipynb`)

```python
from pathlib import Path

project_name = "oi_oro_dental_consolidated"
version_label = "v1"
notebook_file_name = f"{project_name}_{version_label}.ipynb"

print("Notebook dosya adı:", notebook_file_name)
```

---
## 2) Klasör Yapısını Doğrulama ve Çıktı Yolunu Hazırlama

```python
project_root = Path.cwd()
notebook_dir = project_root / "Manuscript_Data" / "03_analysis_scripts"
notebook_dir.mkdir(parents=True, exist_ok=True)

target_notebook_path = notebook_dir / notebook_file_name
print("Hedef notebook yolu:", target_notebook_path)
```

---
## 3) `nbformat` ile Yeni Notebook Nesnesi Oluşturma

```python
import nbformat as nbf

nb = nbf.v4.new_notebook()
nb.metadata = {
    "kernelspec": {
        "display_name": "Python 3",
        "language": "python",
        "name": "python3"
    },
    "language_info": {
        "name": "python",
        "version": "3.13"
    }
}

print("Notebook nesnesi oluşturuldu.")
```

---
## 4) Kod Hücrelerini Programatik Olarak Ekleme

```python
SEED = 20260228

intro_md = """# OI Konsolide Analiz Notebook (v1)\nBu notebook proje kapsamında analiz akışını tek dosyada toplamak için oluşturulmuştur."""
imports_code = """import pandas as pd\nimport numpy as np\n\nnp.random.seed(20260228)"""
example_code = """base = Path.cwd() / 'Manuscript_Data' / '04_final_outputs' / 'tables_csv_and_logs'\n
t1 = pd.read_csv(base / 'publication_table1_overall_FINAL.csv')\nt3 = pd.read_csv(base / 'publication_table3_inferential_FINAL.csv')\n\nprint('Table1 satır sayısı:', len(t1))\nprint('Table3 satır sayısı:', len(t3))\n\nassert len(t1) > 0, 'Table1 boş olamaz'\nassert len(t3) > 0, 'Table3 boş olamaz'"""

nb.cells = [
    nbf.v4.new_markdown_cell(intro_md),
    nbf.v4.new_code_cell(imports_code),
    nbf.v4.new_code_cell(example_code),
]

print('Programatik hücreler notebook nesnesine eklendi.')
```

---
## 5) Notebook’u Diske Yazma ve Dosya Kontrolü

```python
with target_notebook_path.open("w", encoding="utf-8") as f:
    nbf.write(nb, f)

exists_flag = target_notebook_path.exists()
size_bytes = target_notebook_path.stat().st_size if exists_flag else 0

print("Kaydedildi mi?:", exists_flag)
print("Dosya boyutu (bytes):", size_bytes)

assert exists_flag, "Notebook dosyası oluşturulamadı"
assert size_bytes > 0, "Notebook dosyası boş görünüyor"
```

---
## 6) Sürüm Artırma Mantığı (`v1` → `v2`) için Yardımcı Fonksiyon

```python
def next_version_path(base_dir: Path, base_name: str) -> Path:
    """Mevcut dosyalara bakarak _v1, _v2, ... biçiminde çakışmasız isim üretir."""
    version = 1
    while True:
        candidate = base_dir / f"{base_name}_v{version}.ipynb"
        if not candidate.exists():
            return candidate
        version += 1

next_path = next_version_path(notebook_dir, "oi_oro_dental_consolidated")
print("Bir sonraki uygun sürüm adı:", next_path.name)
```
